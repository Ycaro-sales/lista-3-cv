```python
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
def get_good_matches(matches):
    good = []
    for i,(m,n) in enumerate(matches):
        if m.distance < 0.7*n.distance:
          good.append(m)
    return good


def ransac(img1, img2, kp1, kp2, matches, MIN_MATCH_COUNT = 10):
  matchesMask = [[0,0] for i in range(len(matches))]

  good = get_good_matches(matches)

  if len(good)>MIN_MATCH_COUNT:
    src_pts = np.float32([ kp1[m.queryIdx].pt for m in good ]).reshape(-1,1,2)
    dst_pts = np.float32([ kp2[m.trainIdx].pt for m in good ]).reshape(-1,1,2)

    M, mask = cv.findHomography(src_pts, dst_pts, cv.RANSAC,5.0)
    matchesMask = mask.ravel().tolist()

    if M is None:
      return img2, matchesMask, good

    h,w = img1.shape[:2]
    pts = np.float32([ [0,0],[0,h-1],[w-1,h-1],[w-1,0] ]).reshape(-1,1,2)
    dst = cv.perspectiveTransform(pts,M)

    img2_transformed = cv.polylines(img2.copy(),[np.int32(dst)],True,255,3, cv.LINE_AA)
  else:
    matchesMask = None
    img2_transformed = img2

  return img2_transformed, matchesMask, good
```

```python
sift = cv.SIFT_create()
img1 = cv.imread('./assets/1_1.jpeg')
img2 = cv.imread('./assets/1_2.jpeg')
kp1, des1 = sift.detectAndCompute(img1, None)
kp2, des2 = sift.detectAndCompute(img2, None)

FLANN_INDEX_KDTREE = 1
index_params = dict(algorithm=FLANN_INDEX_KDTREE, trees=5)
search_params = dict(checks=50)
flann = cv.FlannBasedMatcher(index_params, search_params)

matches = flann.knnMatch(des1, des2, k=2)
img_out, matchesMask, good = ransac(img1, img2, kp1, kp2, matches, 5)
cv.imshow('img1', img1)
cv.imshow('img_out', img_out)
```
